**Exp3:** Construction of an Inverted Index for a Given Document Collection.

**Aim:**

To construct an inverted index for a collection of text documents and perform basic term-based searching.

**Description:**

1. The program imports modules for file handling and text processing.
2. The preprocess() function converts text into lowercase.
3. Special characters are removed using regular expressions.
4. The cleaned text is split into individual words.
5. The program reads text files from the specified document folder.
6. Only .txt files starting with doc are processed.
7. An inverted index is created using a dictionary of sets.
8. Each term is mapped to the documents containing that term.
9. The program displays the number of documents, vocabulary size, and inverted index.
10. The user can search for a term and view the matching documents and document frequency.

**Source Code:**

````python
import os
import re
from collections import defaultdict

# Function to preprocess a document
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    words = text.split()
    return words

# Function to construct inverted index
def build_inverted_index(document_folder):
    inverted_index = defaultdict(set)
    vocabulary = set()
    documents = sorted(os.listdir(document_folder))
    document_count = 0

    for filename in documents:
        filepath = os.path.join(document_folder, filename)
        if not os.path.isfile(filepath):
            continue

        if not filename.endswith(".txt"): # This will filter out sample_data, etc.
            continue

        # To be robust, also check if filename starts with 'doc' to only process docXX.txt files
        if not filename.startswith('doc'):
            continue

        document_count += 1
        document_id = filename

        with open(filepath, "r", encoding="utf-8") as file:
            text = file.read()

        words = preprocess(text)

        for word in words:
            vocabulary.add(word)
            inverted_index[word].add(document_id)

    return inverted_index, vocabulary, document_count


# Main Program
document_folder = "/content" # Updated path to find the document files
index, vocabulary, document_count = build_inverted_index(document_folder)

print("=" * 60)
print("           INVERTED INDEX CONSTRUCTION")
print("=" * 60)
print("Number of Documents :", document_count)
print("Vocabulary Size     :", len(vocabulary))
print("Number of Terms     :", len(index))
print("=" * 60)

# Display Inverted Index
print("\nINVERTED INDEX")
print("-" * 60)

for term in sorted(index):
    print(f"{term:20} -> {sorted(index[term])}")

# Query Processing
while True:
    query = input("\nEnter search term (or 'exit'): ").lower()

    if query == "exit":
        break

    if query in index:
        documents = sorted(index[query])
        print("\nTerm :", query)
        print("Documents containing the term:")
        print(documents)
        print("Document Frequency :", len(documents))
    else:
        print("\nTerm not found in the collection.")
````
