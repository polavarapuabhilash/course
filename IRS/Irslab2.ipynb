**Exp2:** Pre-processing of a Text Document: Stop Word Removal and Stemming.

**Aim:**

To preprocess a text document by performing tokenization, stop word removal, and stemming using the NLTK library.

**Description:**

1. The program imports the required NLTK modules.

2. The required NLTK resources are downloaded.

3. A sample sentence is given as input text.

4. The text is divided into individual words using tokenization.

5. English stop words are loaded from the NLTK corpus.

6. Common words that do not add significant meaning are removed.

7. Only alphabetic words are retained during preprocessing.

8. The PorterStemmer is used to perform stemming.

9. Each remaining word is reduced to its root or base form.

10. The program displays the original text, filtered words, and stemmed words.

**Source Code:**

````python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab') # Added to download the missing resource
text = "Machine learning algorithm are widely used in data analysis and data science tasks."
words = word_tokenize(text)
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.lower() not in stop_words and word.isalpha()]
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("Original Text:\n", text)
print("\n After Stop Word Removal:\n", filtered_words)
print("After Stemming:\n", stemmed_words)
````
