**Exp1**: Representation of a Text Document in Vector Space Model and Computing Similarity Between Two Documents.

**Aim**:

To implement the cosine Similarity algorithm for measuring the similarity between two text documment in an IRS.

**Description**:

 1)The program uses the TfidfVectorizer from Scikit-learn.

2)Five different text documents are defined as input.

3)English stop words are removed during vectorization.

4)The fit_transform() method converts the documents into a TF-IDF matrix.

5)TF-IDF gives importance to terms based on their frequency in documents.

6)Each document is represented as a numerical vector.

7)The cosine_similarity function is imported from Scikit-learn.

8)The first and second documents are selected for comparison.

9)Cosine similarity measures the similarity between their TF-IDF vectors.

10)The final similarity score is displayed as the output.

````python
from sklearn.feature_extraction.text import TfidfVectorizer

doc1 = "This is the first document."
doc2 = "This document is the second document."
doc3 = "And this is the third one."
doc4 = "Is this the first document?"
doc5 = "The last document is also the fifth document."

vectorizer = TfidfVectorizer(stop_words='english')
Tfid_matrix = vectorizer.fit_transform([doc1, doc2, doc3, doc4, doc5])
````

````python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(Tfid_matrix[0:1], Tfid_matrix[1:2])
print(f"Cosine Similarity: {similarity[0][0]:4f}")
````
